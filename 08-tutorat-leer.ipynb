## Theorie: Dictionaries

### Grundidee

Ein Dictionary speichert **Schlüssel-Wert-Paare**. Über einen Schlüssel greifen wir auf den zugehörigen Wert zu, zum Beispiel über einen Namen auf eine Punktzahl.

```python
punkte = {"Mia": 7, "Noah": 5}
punkte["Mia"]  # ergibt 7
```

Jeder Schlüssel kommt nur einmal vor. Eine erneute Zuweisung an denselben Schlüssel ersetzt den bisherigen Wert. Verschiedene Schlüssel dürfen denselben Wert haben.

### Liste, Tupel oder Dictionary?

| Struktur | Beispiel | Zugriff | Veränderbarkeit |
|---|---|---|---|
| Liste | `[7, 5]` | Über den Index: `werte[0]` | Einträge ändern, hinzufügen und entfernen |
| Tupel | `(7, 5)` | Über den Index: `paar[0]` | Einträge nicht ersetzen, hinzufügen oder entfernen |
| Dictionary | `{"Mia": 7}` | Über den Schlüssel: `punkte["Mia"]` | Paare hinzufügen, Werte ändern und Paare entfernen |

Dictionary-Schlüssel müssen **hashbar** sein. Strings und Zahlen eignen sich als Schlüssel, Listen und Dictionaries nicht. Ein Tupel eignet sich nur, wenn alle seine Elemente hashbar sind. Werte dürfen beliebige Typen haben, auch Listen oder weitere Dictionaries.

### Wichtige Operationen

| Ausdruck | Bedeutung |
|---|---|
| `punkte["Mia"]` | Wert zum Schlüssel lesen; fehlender Schlüssel führt zu `KeyError` |
| `punkte["Mia"] = 9` | Wert ändern oder neues Paar einfügen |
| `"Mia" in punkte` | Prüfen, ob der **Schlüssel** vorhanden ist |
| `punkte.get("Ali")` | Wert lesen; bei fehlendem Schlüssel `None` liefern |
| `punkte.get("Ali", 0)` | Bei fehlendem Schlüssel den Ersatzwert `0` liefern |
| `del punkte["Noah"]` | Paar entfernen; der Schlüssel muss vorhanden sein |
| `len(punkte)` | Anzahl der Schlüssel-Wert-Paare |

### Über ein Dictionary iterieren

- `for name in punkte:` durchläuft die Schlüssel.
- `for wert in punkte.values():` durchläuft die Werte.
- `for name, wert in punkte.items():` durchläuft die Schlüssel-Wert-Paare und entpackt jedes Paar.

Die Klammern bei `values()` und `items()` gehören zum Methodenaufruf. Dictionaries behalten die Einfügereihenfolge bei; die Einträge werden nicht automatisch sortiert. Während der Iteration keine Schlüssel hinzufügen oder entfernen.

### Fehlender Schlüssel: `get` und `None`

Was passiert, wenn wir eine Punktzahl abfragen, die noch nicht gespeichert ist?

1. `punkte.get("Ali")` liefert `None`, weil `"Ali"` hier kein Schlüssel ist.
2. `None` ist ein eigener Python-Wert und steht hier für „kein Eintrag gefunden“.
3. Mit `wert is None` können wir diesen Fall gezielt prüfen.

```python
punkte = {"Mia": 7, "Noah": 5}
wert = punkte.get("Ali")

if wert is None:
    print("Für Ali ist keine Punktzahl gespeichert.")
```

Falls wir stattdessen einen Ersatzwert möchten, schreiben wir `punkte.get("Ali", 0)`. Das ergibt hier `0`. Eine tatsächlich gespeicherte Punktzahl von `0` ist ein gültiger Wert und etwas anderes als `None`.

### Optional: verschachtelte Dictionaries

Ein Dictionary kann weitere Dictionaries oder Listen als Werte enthalten. Den Zugriff können wir in einzelne Schritte zerlegen:

```python
personen = {"Mia": {"alter": 20, "kurse": ["EIDP", "Analysis"]}}

mia = personen["Mia"]        # Dictionary mit Mias Daten
kurse = mia["kurse"]         # Liste ihrer Kurse
erster_kurs = kurse[0]       # "EIDP"
```

Zusammengefasst: `personen["Mia"]["kurse"][0]` liefert ebenfalls `"EIDP"`.


````python
# Dictionary anlegen, verändern und durchlaufen
punkte = {"Mia": 7, "Noah": 5}
punkte["Mia"] = 9
punkte["Lina"] = 7

print("Mias Punktzahl:", punkte["Mia"])
print("Mia als Schlüssel vorhanden:", "Mia" in punkte)

for name, wert in punkte.items():
    print(name, wert)

# Fehlender Schlüssel: get liefert None oder einen gewählten Ersatzwert.
wert = punkte.get("Ali")
print("Ali fehlt:", wert is None)
print("Ersatzwert für Ali:", punkte.get("Ali", 0))

# Eintrag entfernen
del punkte["Noah"]
print("Nach dem Entfernen:", punkte)
````
