## 1 Klassen

### 1.1 Klasse und Instanz

Eine Klasse ist ein Bauplan. Wir bauen uns heute eine Bank, und jede Bank braucht Konten.

```python
class Account:
    pass
```

Aus dem Bauplan können wir so viele Konten erstellen, wie wir wollen. Jedes erstellte Konto ist eine **Instanz** der Klasse.

Konvention: Der Klassenname beginnt mit einem Grossbuchstaben.

```python
konto1 = Account()
konto2 = Account()
```

### 1.2 Attribute

Ein Konto braucht Daten, zum Beispiel einen Inhaber und ein Guthaben. Diese Daten heissen **Attribute**. Wir können sie mit einem Punkt direkt an die Instanz hängen:

```python
konto1.owner = "Mia"
konto1.balance = 100

konto2.owner = "Noah"
konto2.balance = 50

print(konto1.owner, konto1.balance)
```

Das funktioniert, hat aber zwei Probleme:
- Wir schreiben für jedes Konto dasselbe noch einmal.
- Wir können leicht ein Attribut vergessen, und später fehlt dann zum Beispiel das Guthaben.

Besser ist es, schon im Bauplan festzulegen, welche Attribute jedes Konto hat.

### 1.3 `__init__` und `self`

Genau dafür gibt es `__init__`. Diese Methode wird **automatisch aufgerufen**, sobald wir ein neues Konto erstellen.

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
```

`self` ist die Instanz, die gerade erstellt wird. `__init__` nimmt die übergebenen Parameter und speichert sie in dieser Instanz als Attribute.

`self.owner = owner` bedeutet also: „Speichere den übergebenen Namen in diesem Konto.“

Jetzt übergeben wir die Werte direkt beim Erstellen:

```python
konto1 = Account("Mia", 100)
konto2 = Account("Noah", 50)

print(konto1.owner, konto1.balance)
print(konto2.owner, konto2.balance)
```

`self` schreiben wir beim Erstellen nicht hin, das übergibt Python selbst. Wenn wir ein Argument vergessen, meldet Python sofort einen Fehler. So kann kein Attribut mehr fehlen.

```python
konto3 = Account("Lea")
```

### 1.4 Methoden

Bis jetzt kann ein Konto nur Daten speichern. Es soll aber auch etwas können, zum Beispiel Geld einzahlen und abheben. Funktionen, die in einer Klasse stehen, heissen **Methoden**.

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Nicht genug Geld")
        else:
            self.balance = self.balance - amount
```

Auch hier steht `self` an erster Stelle. Die Methode muss wissen, auf welchem Konto sie arbeitet.

```python
konto1 = Account("Mia", 100)
konto1.deposit(50)
konto1.withdraw(30)
print(konto1.balance)

konto1.withdraw(500)
```

Bei `konto1.deposit(50)` setzt Python automatisch `self = konto1`. Darum landet das Geld bei Mia und nicht bei Noah.

Das kennt ihr schon von Listen: `liste.append(3)` ist auch ein Methodenaufruf.

### 1.5 Funktion oder Klasse?

| | Funktion | Klasse |
|---|---|---|
| Was ist es? | eine benannte Folge von Anweisungen | ein eigener Datentyp |
| Fasst zusammen | Anweisungen | Daten (Attribute) und Funktionen (Methoden) |
| Aufruf ergibt | einen Rückgabewert | ein neues Objekt |
| Gedächtnis | keins, nach dem Aufruf ist alles weg | das Objekt behält seine Attribute |
| Schlüsselwort | `def` | `class` |
