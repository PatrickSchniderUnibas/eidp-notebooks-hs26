# Block 1: Boolesche Ausdrücke und Operatoren

## 1.1 Wahrheitswerte

- Neuer Typ: `bool`
- Nur zwei Werte: `True` und `False` (gross geschrieben)
- Ein **boolescher Ausdruck** ergibt einen Wahrheitswert

````python
print(5 > 3)
print(5 < 3)
print(type(5 > 3))

b: bool = 5 > 3
print(b)
````

## 1.2 Vergleichsoperatoren

| Operator | Bedeutung |
|---|---|
| `==` | gleich |
| `!=` | ungleich |
| `<` | kleiner |
| `<=` | kleiner oder gleich |
| `>` | grösser |
| `>=` | grösser oder gleich |

**Achtung:** `=` ist die **Zuweisung**, `==` ist der **Vergleich**.

````python
x = 7
print(x == 7)
print(x != 7)
print(x >= 7)
print(x < 7)
````

## 1.3 Logische Operatoren

`and`, `or`, `not`

| b1 | b2 | `b1 and b2` | `b1 or b2` | `not b1` |
|---|---|---|---|---|
| False | False | False | False | True |
| False | True | False | True | True |
| True | False | False | True | False |
| True | True | True | True | False |

**Rangfolge:** zuerst `not`, dann `and`, zuletzt `or`.
Vergleiche (`<`, `==`, ...) und Rechnen werden **vor** den logischen Operatoren ausgewertet.

**Im Zweifel: Klammern setzen.**

````python
print(True and False)
print(True or False)
print(not True)
````

````python
a = True
b = False
print(not a or b and a)
print((not a) or (b and a))   # gleiche Auswertung, aber lesbarer
print(not (a or b) and a)   
````

### Bedingungen zusammensetzen

Jede Seite von `and` und `or` muss ein **vollständiger Ausdruck** sein.

````python
x = 5
print(x > 0 and x < 10)
print(x < 0 or x > 10)
print(0 < x < 10)   
````

## 1.4 Division mit Rest: `//` und `%`

| Operator | Bedeutung | Beispiel | Ergebnis |
|---|---|---|---|
| `/` | normale Division (immer `float`) | `7 / 2` | `3.5` |
| `//` | ganzzahlige Division (ohne Rest) | `7 // 2` | `3` |
| `%` | Modulo (der Rest) | `7 % 2` | `1` |

Zusammenhang: `a == (a // b) * b + a % b`

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(6 / 3)    # float, auch wenn das Ergebnis "glatt" ist
````

### Beispiel: Uhrzeit
130 Minuten sind wie viele Stunden und Minuten?

````python
minuten = 130
print(minuten // 60, "Stunden und", minuten % 60, "Minuten")
````

### Was kann man damit machen?
- Teilbarkeit prüfen: Rest 0 heisst teilbar
- Zahl zerlegen: `% 10` gibt die letzte Ziffer, `// 10` schneidet sie ab
- Muster, die sich regelmässig wiederholen

````python
print(12 % 4)        # Rest 0, also teilbar
print(13 % 4)
print(4171 % 10)
print(4171 // 10)
````

# Block 2: Die `if`-Anweisung

```
if BEDINGUNG:
    CODE-BLOCK
elif BEDINGUNG:
    CODE-BLOCK
else:
    CODE-BLOCK
```

- **Bedingung**: ein boolescher Ausdruck
- **Doppelpunkt** am Ende der Kopfzeile
- **Block**: Anweisungen mit gleicher **Einrückung** (4 Leerzeichen)
- `True`: Block wird ausgeführt, `False`: Block wird übersprungen

## 2.1 `if` allein
Einen Sonderfall behandeln, danach geht es normal weiter.

````python
a = -3
if a <= 0:
    print("a muss grösser als 0 sein")
    a = 1
print(a)
````

## 2.2 `if` und `else`
Genau zwei Fälle.

````python
a = 5
if a > 5:
    print("a ist grösser als 5")
else:
    print("a ist kleiner oder gleich 5")
````

## 2.3 `if`, `elif`, `else`
Mehrere Fälle. Die Bedingungen werden **von oben nach unten** geprüft.
Es wird **höchstens ein Zweig** ausgeführt (der erste mit `True`).

````python
a = 0
if a < 0:
    print("a ist negativ")
elif a == 0:
    print("a ist gleich 0")
else:
    print("a ist grösser 0")
````

### Wichtiger Unterschied: `elif` oder mehrere `if`?

**Was wird bei `note = 5.5` jeweils ausgegeben?**

````python
note = 5.5

print("Variante A: zwei einzelne if")
if note >= 4:
    print("bestanden")
if note >= 5:
    print("gut")

print("Variante B: if mit elif")
if note >= 5:
    print("gut")
elif note >= 4:
    print("bestanden")
````

# Block 3: Die `while`-Schleife

```
while BEDINGUNG:
    CODE-BLOCK
```

- Bedingung **vor jedem Durchlauf** prüfen
- `True`: Körper ausführen, danach wieder prüfen
- `False`: Schleife ist fertig, das Programm geht nach dem Block weiter
- Ist die Bedingung von Anfang an `False`, läuft der Körper **null Mal**

**Begriffe:** Schleifenbedingung, Schleifenkörper, Durchlauf (Iteration)

## 3.1 Die drei Teile einer Zählschleife
1. **Initialisierung**: Variable vor der Schleife setzen
2. **Bedingung**: wann weiterlaufen?
3. **Veränderung**: im Körper die Variable ändern, sonst endet die Schleife nie

````python
i = 3
while i > 0:
    print(i)
    i = i - 1
print("Start!")
````
