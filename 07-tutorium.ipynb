# 1 Strings

### 1.1 Folge von Zeichen
- String ist wie eine Liste, die Elemente sind Buchstaben
- `len`, Zugriff mit Index ab 0, Leerzeichen zählt mit
- Iteration mit `for c in text`

````python
text = "Hallo World"

print(len(text))     # Länge des Strings mit Leerzeichen

for c in text:       # Zeichen für Zeichen, auch das Leerzeichen
    print(c)
````

```
         H   a   l   l   o
Index:   0   1   2   3   4       
```

````python
s = "Hallo"
print(s[0])                 # erstes Zeichen
print(s[4])                 
print(s[len(s) - 1])        # (letztes Zeichen)
````

### 1.2 Vergleichen
- `==` gleich, `!=` ungleich, Ergebnis ist `True` oder `False`
- Achtung: `=` weist zu, `==` vergleicht
- Gross und Klein sind verschieden, Leerzeichen zählt

````python
print("1:",  "hallo" == "hallo")          
print("2:",  "a" == "A")          
print("3:",  "HalloWorld" == "Hallo World") 
print("4:",  "abc" != "abd")       

s = "Hallo"
print("5:",  s[0] == s[4])         
print("6:",  s[2] == s[3])        

print("7:",  "7" == 7)             
````

- Einzelne Zeichen vergleichen: Man greift mit dem Index auf sie zu
- Der Index fängt bei 0 an, das letzte Zeichen hat `len(text) - 1`


### 1.3 Text ist nicht Zahl
- `"7"` und `7` sind verschiedene Dinge. `"7"` ist **Text** (das Zeichen Sieben), `7` ist eine **Zahl**.
- Python vergleicht nicht, ob es gleich aussieht. Verschiedene Typen sind nie gleich, darum ist `"7" == 7` `False`.
- `int(...)` macht aus Text eine Zahl, `str(...)` aus einer Zahl Text.

````python
print(type("7"))        
print(type(7))        

print("7" == 7)        

print("7" == str(7))     # str( )
print(int("7") == 7)     # int ( ) 
````

- `+` verhält sich bei Text anders als bei Zahlen:

````python
print("7" + "3")              
print(int("7") + int("3"))    
````

### 1.4 Strings verändern
- Strings sind unveränderbar, wie Tupel. Einzelne Zeichen überschreiben geht nicht.
- Diese Zelle gibt absichtlich einen Fehler (`TypeError`):

````python
s = "Hallo"
s[0] = "J"
````

- Man baut stattdessen einen neuen String und verkettet mit `+`. Zwei Strings zusammenfügen heisst Konkatenation:

````python
s = "Hallo"
s = s + "!"                   # "Hallo!"
print(s)
print("Ha" + "llo")           # Hallo
````

- Ein Zeichen in der Mitte ändern: erst mit `list` in eine Liste umwandeln, dort ändern, dann mit `join` wieder zu einem String machen
- Der String vor dem Punkt ist der Trenner, hier ein leerer String (also kein Trenner)

````python
s = "adc"
l = list(s)                   # ["a", "d", "c"]
print(l[1])
l[1] = "b"                    # ["a", "b", "c"]
print("".join(l))             # abc
````

````python
s = "Jallo"
l = list(s)
l[0] = "H"
print("".join(l)) 
````

# 2 Schleifen

### 2.1 Zwei Arten zu schleifen
**Aufgabe:** Gib jedes Zeichen von `"Hallo"` aus. Erst nur die Zeichen, dann mit ihrer Position.
Erwartet: `H a l l o`, danach `0 H`, `1 a`, `2 l`, `3 l`, `4 o`.

**Variante 1: direkt über die Zeichen**
- `for c in text`: in jedem Durchgang ist `c` das nächste Zeichen
- Reicht, wenn man nur die Zeichen braucht

| Durchgang | c |
| --- | --- |
| 1 | H |
| 2 | a |
| 3 | l |
| 4 | l |
| 5 | o |

````python
s = "Hallo"
for c in s:
    print(c)
````

**Variante 2: über die Positionen**
- `for i in range(len(text))`: `i` ist die Position, das Zeichen holt man mit `text[i]`
- Nötig, wenn man die Position braucht

| i | s[i] |
| --- | --- |
| 0 | H |
| 1 | a |
| 2 | l |
| 3 | l |
| 4 | o |

````python
s = "Hallo"

for i in range (len(s)):
    print(i, s[i])
````

**Was macht `range`?**
- `range(start, stop, schritt)` zählt von `start` in Schritten von `schritt` bis **vor** `stop`
- Bei `"Hallo"` ist `len(s)` gleich 5, also läuft `range(5)` von 0 bis 4, genau die Indizes

````python
for i in range(5):
    print(i)                 # 0, 1, 2, 3, 4
````

**Wann welche?**
- Reicht mir das Zeichen: `for c in text`
- Brauche ich die Position: `for i in range(len(text))`


### 2.2 Von hinten zählen
**Aufgabe:** Gib `"Hallo"` Zeichen für Zeichen rückwärts aus. Erwartet: `o l l a H`.

- Letztes Zeichen hat Index `n - 1`, das `i`-te von hinten hat `n - 1 - i`
- `i` läuft vorne von 0 bis `n - 1`, die Position hinten läuft dabei von `n - 1` bis 0

| i | n - 1 - i | Zeichen |
| --- | --- | --- |
| 0 | 4 | o |
| 1 | 3 | l |
| 2 | 2 | l |
| 3 | 1 | a |
| 4 | 0 | H |

````python
s = "Hallo"

n = len (s)

for i in range (n ) :
    print (s[n-1-i])
    
s = "Hallo"

n = len (s)

for j in range (n-1, -1, -1):
    print (s[j])
````

# 3 Prüfungsaufgabe live

Schreiben Sie eine Funktion `longest_word`, welche als Argument einen String mit einem Satz entgegennimmt und das längste Wort in diesem Satz findet und zurückgibt. Wenn mehrere Wörter die gleiche Länge haben, soll das erste zurückgegeben werden.

Als Beispiel soll die Funktion für den Satz `"Der schnelle braune Fuchs"` das Wort `"schnelle"` zurückgeben.

**Vorgehen:**
1. Beispiel von Hand durchgehen: Wörter und Längen aufschreiben
2. Muster erkennen
3. Code schreiben
4. Randfälle testen

````python
# Lösung
def longest_word(sen):
    word = ""
    longest = ""
    
    for c in sen: 
        if c == " " :
            if len(word) > len(longest):
                longest = word
            word = ""
        else: 
            word = word + c
            
    if len( word) > len( longest):
        longest = word
        
    return longest

# Test
longest_word("Der schnelle braune Fuchs")    # "schnelle"
````
