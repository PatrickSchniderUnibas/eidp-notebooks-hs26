# Teil 1: Wiederholung

## 1.1 `print` oder `return`?

- `print` **zeigt** etwas auf dem Bildschirm an. Der Wert ist danach weg.
- `return` **gibt** einen Wert an den Aufrufer **zurück**. Damit kann man weiterrechnen.
- Eine Funktion ohne `return` gibt automatisch `None` zurück.

````python
def anzeigen():
    print(5)
def zurückgeben():
    return 6

x = anzeigen()
y = zurückgeben()
print("x:",x)
print(y)
````

`zeigt_an()` hat zwar 5 auf den Bildschirm geschrieben, aber `x` ist `None`. Nur `gibt_zurueck()` liefert wirklich einen Wert.

**Beides in einer Funktion ist auch möglich:** zuerst etwas ausgeben, dann einen Wert zurückgeben.

````python
def verdoppeln(x):
    print("x wird verdoppelt.")
    return x, 10

print(verdoppeln(5))
````

## 1.2 Verzweigungen: die Reihenfolge zählt

Bei `if / elif / else` wird **von oben nach unten** geprüft. Sobald eine Bedingung wahr ist, wird nur dieser Block ausgeführt, alle weiteren werden übersprungen.

````python
def bewertung(note):
    if note >= 5.5:
        return "sehr gut"
    elif note >= 4: 
        return "bestanden"
    else:
        return "nicht bestanden"
print(bewertung(6))
````

Jetzt vertauschen wir die ersten beiden Bedingungen:

````python
def bewertung(note):
    if note >= 4:
        return "bestanden"
    elif note >= 4: 
        return "sehr gut"
    else:
        return "nicht bestanden"
print(bewertung(6))
````

Die Note 6 ist auch `>= 4`, darum greift schon die erste Bedingung, und `"sehr gut"` wird **nie** erreicht.

**Merke:** Der **speziellste** Fall kommt zuerst, der allgemeinste zuletzt.



# Teil 2: Rekursion

## 2.1 Was ist Rekursion?

**Rekursion** bedeutet: Eine Funktion **ruft sich selbst auf**.
n hinter ihr.

| Baustein | In der Schlange | Im Code |
|---|---|---|
| **Abbruchbedingung** (Basisfall) | Die erste Person muss niemanden fragen | Ein `if`, das **ohne** Selbstaufruf ein Ergebnis liefert |
| **Rekursiver Aufruf** | Die Frage an die Person davor | Die Funktion ruft sich mit einem **kleineren** Problem auf |

## 2.2 Erstes Beispiel: Countdown

````python
def countdown(n):
    if n == 0:
        print("Start")
        return 
    else: 
        print (n)
        countdown( n-1)
countdown(3)
        
````

Was passiert genau?

```
countdown(3)   gibt 3 aus,  ruft countdown(2) auf
countdown(2)   gibt 2 aus,  ruft countdown(1) auf
countdown(1)   gibt 1 aus,  ruft countdown(0) auf
countdown(0)   Abbruchbedingung erfüllt: gibt "Start!" aus, kein weiterer Aufruf
```

Bei jedem Aufruf wird `n` um 1 **kleiner**. Irgendwann ist `n == 0`, und die Rekursion hört auf.

## 2.3 Was passiert ohne Abbruchbedingung?

Wenn die Abbruchbedingung fehlt, ruft sich die Funktion **endlos** selbst auf. Python bricht nach etwa 1000 Aufrufen mit einem **`RecursionError`** ab.

```python
def countdown_kaputt(n):
    print(n)
    countdown_kaputt(n - 1)    # keine Abbruchbedingung!

countdown_kaputt(3)
```

Ausgabe (gekürzt):

```
3
2
1
0
-1
-2
...
-994
RecursionError: maximum recursion depth exceeded
```

Die Zahlen laufen ins Negative weiter, weil nichts die Rekursion stoppt.

Dasselbe passiert, wenn das Problem **nicht kleiner** wird, z. B. bei `countdown(n)` statt `countdown(n - 1)`, oder wenn die Abbruchbedingung **nie erreicht** wird, z. B. `countdown(-1)` mit der Abbruchbedingung `n == 0`.

**Die Abbruchbedingung ist das Wichtigste an jeder Rekursion.** Darum schreibt man sie immer **zuerst**.

## 2.4 Rekursion mit Rückgabewert: Summe von 1 bis n

Die Summe `1 + 2 + ... + n` kann man so beschreiben:

- Die Summe bis 0 ist **0**. (Abbruchbedingung)
- Die Summe bis n ist **n + Summe bis n-1**. (rekursiver Fall)

````python
# Beispiel: summe(n)
def summe(n):
    if n == 0:
        return 0
    return n + summe(n-1)

summe(4)
````

Der Ablauf hat **zwei Richtungen**:

**Hinweg** (Aufrufe gehen nach unten, bis die Abbruchbedingung erreicht ist):

```
summe(4) = 4 + summe(3)
               summe(3) = 3 + summe(2)
                              summe(2) = 2 + summe(1)
                                             summe(1) = 1 + summe(0)
                                                            summe(0) = 0
```

**Rückweg** (Ergebnisse werden nach oben zurückgegeben und zusammengerechnet):

```
                                                            summe(0) gibt 0 zurück
                                             summe(1) gibt 1 + 0 = 1 zurück
                              summe(2) gibt 2 + 1 = 3 zurück
               summe(3) gibt 3 + 3 = 6 zurück
summe(4) gibt 4 + 6 = 10 zurück
```

# Prüfungsaufgabe: Fibonacci

Diese Aufgabe kam in der Prüfung vom 14. Januar 2025 als Aufgabe 13 (3 Punkte):

> Schreiben Sie eine rekursive Funktion `fibonacci`, welche als Argument eine ganze Zahl `n` erhält und die `n`-te Fibonacci-Zahl f<sub>n</sub> zurückgibt. Die Fibonacci-Zahlen sind wie folgt definiert:
>
> - f<sub>0</sub> = 1
> - f<sub>1</sub> = 1
> - f<sub>k</sub> = f<sub>k-1</sub> + f<sub>k-2</sub>

Die Folge beginnt so: **1, 1, 2, 3, 5, 8, 13, 21, ...** Jede Zahl ist die Summe der zwei Zahlen davor.

Wir gehen wieder nach dem Rezept vor:

**1. Abbruchbedingung:** Wo steht in der Definition ein Ergebnis, ohne dass man etwas ausrechnen muss?

**2. Problem verkleinern:** Welche kleineren Probleme brauchen wir für f<sub>n</sub>?

**3. Zusammensetzen:** Wie kommen wir von den kleineren Ergebnissen zu f<sub>n</sub>?

````python
# Prüfungsaufgabe: fibonacci(n)
def fibonacci(n):
    if n == 0 or n == 1:
        return 1
    return fibonacci(n-1)  + fibonacci(n-2)

print(fibonacci(4))
print(fibonacci(6))
````
