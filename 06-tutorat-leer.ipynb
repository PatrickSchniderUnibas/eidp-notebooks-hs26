### For Loops:

Wichtig zu verstehen: **For-Schleifen ermöglichen es, über die Elemente einer Sequenz zu iterieren**, ohne manuell einen Index verwalten zu müssen.

Im Gegensatz zu While-Schleifen eignen sie sich besonders, wenn eine Sequenz durchlaufen oder eine bestimmte Anzahl Wiederholungen ausgeführt werden soll.

**1. Direkt über die Elemente einer Liste / eines Tupels iterieren:**
```python
for element in liste:
    print(element)
```





#### ```for ... in ...:```als Python Keywords

````python
## For loops
l = [0, 2, 4, 6, 8]
for elem in l:
    print("hello", elem)
````


**2. Mit range() über einen Zahlenbereich iterieren:**
```python
for i in range(start, stop, schrittgrösse):
    print(i)
```

While loops:

For loops:

## Datenstrukturen

### Tupel ( )

Wichtig zu verstehen: **Sequenz fester Länge**, die einzelnen Elemente können von unterschiedlichen Datentypen sein.

#### Syntax:

Die für Tupel verwendete Syntax besteht aus ```tupelVariableName = ( ... )```

Um auf ein Element eines Tupels zu**zugreifen**, werden **eckige Klammern** mit einem Index verwendet. ```name_tupel[index_pos]```

Die Elemente von Tupeln kann man nicht direkt untereinander wechseln -> Tupel sind immutable (unveränderlich)
Um einem Tupel einen neuen Wert zuzuweisen, muss das gesamte Tupel neu zugewiesen werden.



### Tuple Immutable (unveränderlich)

````python
tu = ("hello", 3)
print(tu)

## tu[0], tu[1] = tu[1], tu[0]

## tu = (tu[1], tu[0])

print(tu)
````

### Listen [ ]

Wichtig zu verstehen: **Sequenz variabler Länge**, dabei sind alle Elemente üblicherweise vom selben Datentyps. Es ist möglich Listen bestehend aus mehreren Datentypen zu erstellen.

Man kann schnell und einfach zwei Elemente einer Liste wechseln -> Listen sind **mutable (veränderlich)**

#### Syntax:
Die für Listen verwendete Syntax ist: ```listVariableName = [ ... ]```. Um auf das i. Element der Liste zuzugreifen, verwenden wir ```listVariableName[i]```

Um listen zu verändern, stehen uns einige nützliche Operationen zur Verfügung

- list_name.append(what)
- list_name.insert(index, what)
- list_name.pop(index)
- list_name1 + list_name2
- list_name * number
- sorted(list_name)
- "element" in list_name -> boolean 

### Lists mutable (veränderlich)

````python
li = ["hello", 5]
print(li)
li[0], li[1] = li[1], li[0]
print(li)
````

````python
## List append
l = []
for i in range(0,10,2):
    l.append(i)
print(l)
````

````python
## List insert
l = [0, 2, 4, 6, 8]
n = len(l)
for i in range(0,n):
    l.insert(i*2, "-")
print(l)
````

**Was geschieht hier?**

````python
## Pop element - 
import math
new_l = list(range(0,30))
for i in range(0,len(new_l)-1):
    if math.sqrt(new_l[i])%1 != 0:
        new_l.pop(i)
print(new_l)
````

````python
l = ["foo", "bar"]*3
print(l)
l.pop(3)
print(l)
l.pop(3)
print(l)
````

### Type Annotation

**Wieso?**

Type Annotations ermöglichen es, die erwarteten Datentypen von Variablen, Funktionsparametern und Rückgabewerten anzugeben.

- **Lesbarkeit:** Der Code wird verständlicher, besonders bei grösseren Programmen.
- **Fehlererkennung:** Entwicklungsumgebungen und externe Tools können mögliche Typfehler frühzeitig erkennen.

**Aber:** Type Annotations sind in Python optional und werden zur Laufzeit nicht automatisch überprüft. Sie schränken die tatsächlich verwendbaren Datentypen also nicht ein.

````python
def something(n: str) -> None:
    return(n)
````

````python
print(something(123)) ## Funktioniert torzdem 
````
